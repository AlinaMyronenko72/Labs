```python
import sys
def sort(arr):
    for i in range(0, len(arr)):
        for j in range(0, len(arr) - i - 1):
            if arr[j] < arr[j + 1]:
                temp = arr[j]
                arr[j] = arr[j + 1]
                arr[j + 1] = temp
                
def simp(arr):
    i=0
    arr2=[]
    while i<len(arr):
        if arr[i]<10:
            if arr[i]==4 or arr[i]==6 or arr[i]==8 or arr[i]==9:
                 arr2.append(arr[i])
        else:
            if arr[i]%2==0 or arr[i]%3==0 or arr[i]%5==0 or arr[i]%7==0:
                arr2.append(arr[i])
        i=i+1
    return arr2 
def task9():
    try:    
        l = list(map(int,input('Enter value: ').split())) 
    except ValueError:
        print("You enter somethig wrong!")
        sys.exit(0)
    l2=simp(l)
    sort(l2)
    print("Sorted list:")
    print(l2)


```
