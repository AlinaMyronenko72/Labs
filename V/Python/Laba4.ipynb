```python
import numpy as np
import csv
```

1. Створіть функцію, яка приймає назву файлу та назву стовпця як
параметр та повертає масив numpy зі значеннями стовпця.

```python
def read(file, name):
    with open(file, 'r') as f:
        read = csv.reader(f)
        mas = list(read)
        i=mas[0].index(name)
        res=np.array(mas[1:])[:, i]
    return res
```

2. Використовуючи функцію з попереднього завдання, створіть масив
numpy зі значень стовпця «APC» (Armored personnel carrier).

```python
APC=read('D:\\Учеба\\9 семестр\\Python\\Laba4\\russia_losses_equipment.csv','APC')
APC = APC.astype("int")
print(APC)
```

3. Знайдіть добові втрати БТР.

```python
def lost_APC(mas):
    res=np.diff(mas,prepend=mas[0])
    res[0]=mas[0]
    return res
losts=lost_APC(APC)
print(losts)
```

4. Знайдіть 5 найбільших добових втрат БТР.

```python
def top5(mas):
    sort=np.sort(mas)[::-1]
    top=sort[:5]
    return top
top=top5(losts)
print(top)
```

5. Визначте, скільки БТР було знищено влітку 2023 року.

```python
date=read('D:\\Учеба\\9 семестр\\Python\\Laba4\\russia_losses_equipment.csv','date')
date=date.astype(str)
def des(mas,date):
    m = ((date >= "2023-06-01") & (date <= "2023-08-31"))
    des=np.sum(mas[m])
    return des
des=des(losts,date)
print(des)
```

6. Знайдіть середнє значення знищених БТР за 100-500 днів війни.

```python
def average(mas):
    temp=mas[99:500]
    av = np.mean(temp)
    return av
average=average(losts)
print(average)
```
