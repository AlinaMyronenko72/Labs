```python
14. Знайти рік із найбільшою кількістю гендерно нейтральних імен (однакові
чоловічі та жіночі імена)
```

```python
group = df.groupby(['Year', 'Name', 'Gender']).size().unstack()
gender = (group['F'] > 0) & (group['M'] > 0)
sumy = gender.groupby('Year').sum()
year = sumy.idxmax()
count = sumy.max()
res = pd.DataFrame({'Year': [year], 'Gender_neutral_names': [count]})
res
```

```python
22. Знайти найпопулярніші гендерно-нейтральні імена (ті, що присутні
кожного року)
```

```python
names = {}
M = df[df['Gender'] == 'M'].groupby('Year')['Name'].unique()
F = df[df['Gender'] == 'F'].groupby('Year')['Name'].unique()
for name in df['Year'].unique():
    names[name] = set(M[name]) & set(F[name])
all = set.intersection(*names.values())
popular = pd.DataFrame({'0': list(all)})
popular
```
